```python
# Import Libraries

from pyspark.sql.functions import *
```

```python
# Load All Silver Tables 
customers = spark.table("silver_customers")

billing = spark.table("silver_billing")

usage = spark.table("silver_usage")

support = spark.table("silver_support")

churn = spark.table("silver_churn")
```

```python
# Verify Tables 
customers.count()

billing.count()

usage.count()

support.count()

churn.count()
```

```python
# Check Schema 
customers.printSchema()

billing.printSchema()

usage.printSchema()

support.printSchema()

churn.printSchema()
```

```python
# Join Customers + Billing 
customer_360 = customers.join(

    billing,

    on="CustomerID",

    how="left"

)
```

```python
display(customer_360.limit(5))
```

```python
# Join Usage 
customer_360 = customer_360.join(

    usage,

    on="CustomerID",

    how="left"

)
```

```python
display(customer_360.limit(5))
```

```python
# Join Support 
customer_360 = customer_360.join(

    support,

    on="CustomerID",

    how="left"

)
```

```python
display(customer_360.limit(5))
```

```python
# Join Churn 
customer_360 = customer_360.join(

    churn,

    on="CustomerID",

    how="left"

)
```

```python
# Check Record Count 
customer_360.count()
```

```python
# Verify CustomerID Uniqueness 
customer_360.select(

    "CustomerID"

).distinct().count()
```

```python
# Handle Missing Values 
customer_360 = customer_360.fillna({

    "MonthlyRevenue":0,

    "TransactionCount":0,

    "TotalLogins":0,

    "LifetimeMinutes":0,

    "AvgSession":0,

    "ActiveDays":0,

    "LoginFrequency":0,

    "RecentlyActive":0,

    "TotalTickets":0,

    "OpenTickets":0,

    "ClosedTickets":0,

    "ComplaintScore":0,

    "SupportLevel":"LOW",

    "SupportRisk":"LOW",

    "Churn":0,

    "ChurnStatus":"ACTIVE",

    "RetentionFlag":1

})
```

```python
display(customer_360.limit(5))
```

```python
# Validate Null Values 
customer_360.select([

count(

when(col(c).isNull(),c)

).alias(c)

for c in customer_360.columns

]).show()
```

```python
# Verify Schema 
customer_360.printSchema()
```

```python
# Preview Final Table 
display(customer_360.limit(20))
```

```python
# Customer Statistics 
customer_360.describe().show()
```

```python
# Save Customer 360 
customer_360.write.mode(

"overwrite"

).saveAsTable(

"silver_customer_360"

)
```

```python
# Validate Saved Table 
display(

spark.table(

"silver_customer_360"

)
)
```
