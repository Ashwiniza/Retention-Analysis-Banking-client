```python
# Prepare the final analytical dataset that Power BI will consume. 
# Import Libraries 
from pyspark.sql.functions import *
```

```python
# Load Gold Table
gold = spark.table("gold_retention_features")
```

```python
# Verify Data 
display(gold.limit(10))
```

```python
gold.printSchema()
```

```python
print("Total Records:", gold.count())
```

```python
from pyspark.sql.functions import (
    col,
    count,
    when,
    sum,
    round,
    avg,
    max,
    min
)
```

```python
# Check Missing Values 
gold.select([
    count(when(col(c).isNull(), c)).alias(c)
    for c in gold.columns
]).show()
```

```python
# Select Business Columns 
powerbi_dataset = gold.select(

    "CustomerID",

    "JoinDate",

    "Region",

    "Segment",

    "SubscriptionType",

    "Age",

    "AgeGroup",

    "TenureDays",

    "BillingMonth",

    "MonthlyRevenue",

    "TransactionCount",

    "RevenuePerLogin",

    "TotalLogins",

    "LifetimeMinutes",

    "AvgSession",

    "LoginFrequency",

    "UsageLevel",

    "SessionCategory",

    "RecentlyActive",

    "TotalTickets",

    "OpenTickets",

    "ClosedTickets",

    "ComplaintScore",

    "SupportLevel",

    "SupportRisk",

    "OpenTicketRatio",

    "EngagementScore",

    "ComplaintRatio",

    "RevenueCategory",

    "LifetimeSegment",

    "LoginCategory",

    "UsageCategory",

    "RiskScore",

    "HighRiskCustomer",

    "CustomerHealth",

    "HealthCategory",

    "LoyaltyScore",

    "Churn",

    "ChurnStatus",

    "RetentionFlag"

)
```

```python
# Verify Dataset 
display(powerbi_dataset)
```

```python
# Validate Total Records 
print("Rows :", powerbi_dataset.count())
```

```python
# Validate Customer Count 
print(
    "Unique Customers:",
    powerbi_dataset.select("CustomerID").distinct().count()
)
```

```python
from pyspark.sql.functions import sum, round

powerbi_dataset.groupBy("Region") \
.agg(
    round(
        sum("MonthlyRevenue"), 2
    ).alias("Revenue")
).show()
```

```python
# Churn Summary 
powerbi_dataset.groupBy("ChurnStatus").count().show()
```

```python
# Revenue by Segment 
powerbi_dataset.groupBy("Segment")\
.agg(
    round(
        sum("Monthly Revenue"),2
    ).alias("Revenue")
).show()
```

```python
# Revenue Category 
powerbi_dataset.groupBy(
    "Revenue Category"
).count().show()
```

```python
# Customer Health 
powerbi_dataset.groupBy(
    "Health Category"
).count().show()
```

```python
# High Risk Customers 
powerbi_dataset.filter(
    col("High Risk Customer")==1
).show()
```

```python
# Recently Active Customers 
powerbi_dataset.groupBy(
    "Recently Active"
).count().show()
```
